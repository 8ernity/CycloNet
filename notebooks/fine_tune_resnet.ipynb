# Fine-Tune ResNet50 for Cyclone Intensity Classification
This notebook demonstrates how to load a pre-trained ResNet50 backbone (which currently powers the backend) and fine-tune it on cyclone-specific data (like the Digital Typhoon dataset) to accurately output the 7 IMD categories.

```python
import torch
import torch.nn as nn
from torchvision.models import resnet50, ResNet50_Weights
import torch.optim as optim

# 1. Load the pre-trained ResNet50
model = resnet50(weights=ResNet50_Weights.DEFAULT)

# 2. Freeze the base layers (optional, but good for starting transfer learning)
for param in model.parameters():
    param.requires_grad = False

# 3. Replace the final fully connected layer to output our 7 IMD categories
num_ftrs = model.fc.in_features
model.fc = nn.Linear(num_ftrs, 7) # 7 classes: D, DD, CS, SCS, VSCS, ESCS, SuCS

device = torch.device("cuda:0" if torch.cuda.is_available() else "cpu")
model = model.to(device)

print("Model architecture ready for 7-class cyclone classification.")

# 4. Setup Optimizer & Loss
criterion = nn.CrossEntropyLoss()
optimizer = optim.Adam(model.fc.parameters(), lr=0.001)

# Note: You will need to implement a PyTorch DataLoader here pointing to your image directory
# dataloaders = { 'train': train_loader, 'val': val_loader }

def train_model(model, criterion, optimizer, num_epochs=10):
    print("Training loop would run here with a proper dataloader.")
    # for epoch in range(num_epochs):
    #     for inputs, labels in dataloaders['train']:
    #         inputs, labels = inputs.to(device), labels.to(device)
    #         optimizer.zero_grad()
    #         outputs = model(inputs)
    #         loss = criterion(outputs, labels)
    #         loss.backward()
    #         optimizer.step()
    return model

# To save the weights for the FastAPI backend:
# torch.save(model.state_dict(), '../backend/app/models/cyclone_resnet_v1.pth')
```
